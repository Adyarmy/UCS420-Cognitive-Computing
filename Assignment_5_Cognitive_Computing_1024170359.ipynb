## Q.1 Basic NumPy Array Operations

Write a program to create a NumPy 1D-array with 5 elements and perform the following basic operations:
- Addition of 2 to all elements
- Multiplication of all elements by 3
- Division of every element by 2

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print("Original array:", arr)
print("After adding 2:", arr + 2)
print("After multiplying by 3:", arr * 3)
print("After dividing by 2:", arr / 2)
```

### b) Find the most frequent value and its index/indices

Find the most frequent value and its indices for the given arrays `x` and `y`.

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

def most_frequent_value_and_indices(arr):
    values, counts = np.unique(arr, return_counts=True)
    max_count = counts.max()
    most_frequent = values[counts == max_count]

    print("Array:", arr)
    print("Most frequent value(s):", most_frequent)
    for value in most_frequent:
        print(f"Indices of {value}:", np.where(arr == value)[0])

print("For x:")
most_frequent_value_and_indices(x)

print("\nFor y:")
most_frequent_value_and_indices(y)
```

## Q.5 NumPy 2-D Array Operations

Create a 2-D array of three rows and four columns named `ucs420_Adarsh` using the given values. Compute its mean, median, maximum, minimum, and unique elements. Then reshape it to four rows and three columns as `reshaped_ucs420_Adarsh` and resize it to two rows and three columns as `resized_ucs420_Adarsh`.

```python
ucs420_Adarsh = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:")
print(ucs420_Adarsh)

print("\nMean:", np.mean(ucs420_Adarsh))
print("Median:", np.median(ucs420_Adarsh))
print("Maximum:", np.max(ucs420_Adarsh))
print("Minimum:", np.min(ucs420_Adarsh))
print("Unique elements:", np.unique(ucs420_Adarsh))

reshaped_ucs420_Adarsh = ucs420_Adarsh.reshape(4, 3)
print("\nReshaped array (4 x 3):")
print(reshaped_ucs420_Adarsh)

resized_ucs420_Adarsh = np.resize(ucs420_Adarsh, (2, 3))
print("\nResized array (2 x 3):")
print(resized_ucs420_Adarsh)
```

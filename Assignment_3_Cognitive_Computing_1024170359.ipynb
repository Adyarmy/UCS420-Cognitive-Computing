## Q.6 Create a sample dataset (`employees.csv`) containing information about employees in a company.

Then perform all the given operations from parts (a) to (l).

```python
employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(employee_data)
employees.to_csv("employees.csv", index=False)
employees
```

### Q.6 (e) Calculate the following statistics from the dataset.

i. The average salary of employees.  
ii. The total bonus paid to all employees.  
iii. The youngest employee's age.  
iv. The highest performance rating.

```python
print("i. Average salary:", employees["Salary"].mean())
print("ii. Total bonus:", employees["Bonus"].sum())
print("iii. Youngest employee's age:", employees["Age"].min())
print("iv. Highest performance rating:", employees["Rating"].max())
```

### Q.6 (f) Sort the DataFrame by the Salary column in descending order.

```python
employees_sorted = employees.sort_values(by="Salary", ascending=False)
employees_sorted
```

### Q.6 (g) Add a new column that categorizes employees based on their performance rating.

- Excellent for ratings >= 4.5
- Good for ratings >= 4.0 but < 4.5
- Average for ratings < 4.0

```python
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance_Category"] = employees["Rating"].apply(performance_category)
employees
```

### Q.6 (h) Identify missing values in the DataFrame.

### Q.6 (i) Rename the Employee_ID column to ID.

```python
employees.rename(columns={"Employee_ID": "ID"}, inplace=True)
employees
```

### Q.6 (j) Find all employees who:

i. Have more than 5 years of experience.  
ii. Belong to the IT department.

```python
print("i. Employees with more than 5 years of experience:")
print(employees[employees["Years_of_Experience"] > 5])

print("\nii. Employees belonging to the IT department:")
print(employees[employees["Department"] == "IT"])
```

## Q.1 Sensor Readings and Vectorization

Consider the given sensor readings. The sensor has an average measurement error of +2°C. Perform the required operations without using a `for` loop.

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Add +2°C to every reading using vectorization
corrected_temperature = temperature + 2
print("Corrected temperatures:", corrected_temperature)

# b. Convert the corrected temperatures from Celsius to Fahrenheit
fahrenheit = (9/5) * corrected_temperature + 32
print("Temperatures in Fahrenheit:", fahrenheit)

# c. Identify all readings greater than 32°C using Boolean indexing
readings_greater_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", readings_greater_32)

# d. Count how many readings exceed 32°C
count_greater_32 = np.sum(corrected_temperature > 32)
print("Number of readings exceeding 32°C:", count_greater_32)

# e. Explanation:
# Vectorization performs an operation on the complete NumPy array at once,
# avoiding an explicit Python loop. Boolean indexing creates a Boolean mask
# and directly selects only the values satisfying a condition. These operations
# are concise and generally faster and more efficient than processing each
# element individually with a Python for loop.
```

## Q.2 Steps Matrix and Aggregation

The matrix contains the daily step counts for four users over three days. Use NumPy aggregation functions and the appropriate axes.

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Calculate the total steps recorded
total_steps = np.sum(steps)
print("Total steps:", total_steps)

# b. Calculate the mean number of steps
mean_steps = np.mean(steps)
print("Mean steps:", mean_steps)

# c. Find the maximum and minimum values
maximum_steps = np.max(steps)
minimum_steps = np.min(steps)
print("Maximum steps:", maximum_steps)
print("Minimum steps:", minimum_steps)

# d. Total steps for each day (axis=0)
daily_totals = np.sum(steps, axis=0)
print("Total steps for each day:", daily_totals)

# e. Total steps for each user (axis=1)
user_totals = np.sum(steps, axis=1)
print("Total steps for each user:", user_totals)

# f. User/day position containing the maximum number of steps using argmax()
max_flat_index = np.argmax(steps)
max_position = np.unravel_index(max_flat_index, steps.shape)
print("Maximum value:", steps[max_position])
print("User/day position (0-based index):", max_position)
```

## Q.3 NumPy Arrays, Slicing, Copying, Reshaping and Flattening

Write a NumPy program demonstrating slicing, the difference between views and copies, reshaping, indexing, `flatten()`, `ravel()`, and array properties.

```python
# a. Create the original NumPy array
original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)

# b. Slice elements from index 1 to 4
subset = original[1:5]
print("Subset:", subset)

# c. Modify the first element of subset to 999
# A normal NumPy slice is a view, so the original is also affected.
subset[0] = 999
print("Original after modifying subset:", original)
print("Subset after modification:", subset)

# d. Create another slice using .copy() and modify its first element to 500
copied_array = original[1:5].copy()
copied_array[0] = 500
print("Original after modifying copied array:", original)
print("Copied array:", copied_array)

# e. Create numbers 1 to 12 and reshape into a 3 x 4 matrix
matrix = np.arange(1, 13).reshape(3, 4)
print("3 x 4 matrix:")
print(matrix)

# f. Indexing and slicing
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Elements from rows 1-2 and columns 2-3:")
print(matrix[1:3, 2:4])

# g. Flatten using flatten() and ravel()
flattened = matrix.flatten()
raveled = matrix.ravel()
print("Using flatten():", flattened)
print("Using ravel():", raveled)

# h. Modify an element returned by ravel()
raveled[0] = 100
print("After modifying ravel():")
print("Raveled array:", raveled)
print("Original matrix:", matrix)
print("The original matrix is affected because ravel() normally returns a view.")

# i. Modify an element returned by flatten()
flattened[1] = 200
print("After modifying flatten():")
print("Flattened array:", flattened)
print("Original matrix:", matrix)
print("The original matrix is not affected because flatten() returns a copy.")

# j. Display shape, ndim, size and dtype
print("Shape:", matrix.shape)
print("Number of dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

## Q.4 Cognitive-Assistive System Using Ordinary Least Squares

Use NumPy to calculate the transpose, matrix product, inverse, OLS coefficients, and the predicted assistance score for a new user.

```python
# Given data
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a. Display the shape and dimensions of X
print("Shape of X:", X.shape)
print("Dimensions (ndim) of X:", X.ndim)

# b. Calculate X.T
X_T = X.T
print("X.T:")
print(X_T)

# The transpose changes rows into columns and columns into rows.
# Thus, the three feature columns of X become three rows in X.T.

# c. Calculate X.T @ X
XTX = X_T @ X
print("X.T @ X:")
print(XTX)

# d. Calculate the matrix inverse
XTX_inverse = np.linalg.inv(XTX)
print("Inverse of X.T @ X:")
print(XTX_inverse)

# e. Implement the Ordinary Least Squares equation:
# beta = (X.T @ X)^(-1) @ X.T @ y
beta = XTX_inverse @ X_T @ y
print("OLS coefficients (beta):")
print(beta)

# f. Interpretation:
# beta[0] is the coefficient for sleep hours.
# beta[1] is the coefficient for activity level.
# beta[2] is the coefficient for stress level.
# Each coefficient indicates the estimated change in assistance score
# associated with a one-unit change in that feature, while the other
# features are held constant.

# g. Predict the assistance score for a new user
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("New user:", new_user)
print("Predicted assistance score:", predicted_score)
```
